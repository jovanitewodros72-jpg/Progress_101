```python
To_do_list = []
completed_task = []

def add_function():
    add_task = input("Type the task you want to add: ")
    To_do_list.append(add_task)

def remove_function():
    remove_task = input("Type the task you want to remove: ")
    if remove_task in To_do_list:
        To_do_list.remove(remove_task)
    else:
        print("Task is not found. Try again!")

def edit_function():
    original_task = input("Type the task you want to edit first: ")
    if original_task in To_do_list:
        edited_task = input("Type the task you want to replace it with: ")
        To_do_list.remove(original_task)
        To_do_list.append(edited_task)
    else:
        print("Task is not found. Try again!")

def exit_function():
    print("Thank you!")
while True:
    print("--- TO-DO-LIST ---")
    print(To_do_list)
    print("--- COMPLETED-TASKS---")
    print(completed_task)
    print("=== MAIN MENU === \n 1. Add New Task \n 2. Remove Task \n 3. Edit Task \n 4. Mark Task as Completed \n 5. Exit")

    number_choice = str(input("Enter your choice (1-5): "))

    if number_choice == "1":
        add_function()

    elif number_choice == "2":
        remove_function()

    elif number_choice == "3":
        edit_function()

    elif number_choice == "4":
        task_done = input("tell me a task you completed: ")
        for task_done in To_do_list:
            To_do_list.remove(task_done)
            completed_task.append(task_done)

    elif number_choice == "5":
        exit_function()
        break

    else:
        print("Invalid input! Enter a number from 1 - 5! ")
```
